# Feature engineering
Feature engineering is the process of using domain knowledge to transform raw data into informative features (variables) that enhance machine learning model performance. It involves creating, selecting, and modifying inputs—such as scaling, encoding, or aggregating data—to help algorithms detect patterns more effectively, ultimately increasing predictive accuracy.

```python
# import data and necessary libraries
import pandas as pd

df = pd.read_csv('../data/processed/food_delivery.csv')
df.head()
```

```python
# delivery_delay_gap
# delivery_time_minutes – estimated_delivery_time = late, negative = early.

df['delivery_delay_gap'] = df['delivery_time_minutes'] - df['estimated_delivery_time']
```

```python
# is peak hour
# Binary: 1 if order_hour in {12, 13, 14, 20, 21, 22, 23, 0}. Lunch and dinner rush
df['is_peak_hour'] = df['order_hour'].isin([12, 13, 14, 20, 21, 22, 23, 0])
```

```python
# cost per km
df['cost_per_km'] = df['delivery_fee'] / df['delivery_distance_km']
```

```python
# revenue per item
df['revenue_per_item'] = df['order_value'] / df['number_of_items']
```

```python
# is night order
# Binary: 1 if order_hour in {0, 1, 2, 3, 4, 5}. Late night orders
df['is_night_order'] = df['order_hour'].isin([22 ,23 ,0, 1, 2, 3, 4, 5])
```
