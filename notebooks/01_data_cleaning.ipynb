# Data cleaning
Data cleaning is a process in data analysis of removing unwanted, non-logic records, missing empty values and prepare a cleaned data for analysis

## About dataset

```python
# About data

import pandas as pd

# Display all columns
pd.set_option('display.max_columns', None)

# Optional: widen display
pd.set_option('display.width', 1000)

# Load dataset
df = pd.read_csv("../data/raw/food_delivery.csv")

# Dataset shape
print("Dataset shape:", df.shape)

# First rows with all columns visible
print("\nFirst 5 rows:")
print(df.head())

# Descriptive statistics
print("\nDataset statistics:")
print(df.describe(include='all'))
```

```python
# check null values

print("\nNull values per column:")
print(df.isnull().sum())
```

```python
# Display all rows with null values

df[df.isna().any(axis=1)]
```

```python
print("NULL values in delivery_partner_rating:")
df[df['delivery_partner_rating'].isnull()][
    ['cancellation_flag', 'refund_flag']
].value_counts()
```

```python
print("NULL values in customer_rating:")
df[df['customer_rating'].isnull()][
    ['cancellation_flag', 'refund_flag']
].value_counts()
```

```python
print("NULL values in delivery_partner_rating:")
df[df['delivery_partner_rating'].isnull()][
    ['cancellation_flag', 'refund_flag']
].value_counts()
```

```python
# Preserve missing information
df['customer_rating_missing'] = df['customer_rating'].isnull().astype(int)
df['delivery_partner_rating_missing'] = df['delivery_partner_rating'].isnull().astype(int)

# Fill ratings with median
df['customer_rating'] = df['customer_rating'].fillna(df['customer_rating'].median())
df['delivery_partner_rating'] = df['delivery_partner_rating'].fillna(df['delivery_partner_rating'].median())

# Missing tip = no tip
df['tip_amount'] = df['tip_amount'].fillna(0)

# check null null values again
print("\nNull values per column after imputation:")
print(df.isnull().sum())
```

```python
# Check for duplicates
print("\nNumber of duplicate rows:", df.duplicated().sum())
```
